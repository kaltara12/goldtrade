# **Note**

Kode di atas menarik modul Image serta ImageEnhance dari library Pillow, kemudian mengakses dan membuka berkas gambar yang tersimpan di Google Drive.

```python
from PIL import Image
from PIL import ImageEnhance

# Opens the image file
image = '/content/drive/MyDrive/CV/cabe.jpg'
image = Image.open(image)
# shows image in image viewer
#img = plt. imread('image.png')
plt.imshow(image)
#image.show()


# Enhance Brightness
```

# **Catatan**

Penyesuaian Kecerahan (Brightness): Nilai variabel new_bri ditetapkan sebesar 0,5. Karena angka ini kurang dari 1,0, tingkat kecerahan visual pohon cabe mengalami penurunan sehingga tampak lebih redup dari kondisi semula.

```python
curr_bri = ImageEnhance.Brightness(image)
new_bri = 0.5

# Brightness enhanced by a factor of 2.5
img_brightened = curr_bri.enhance(new_bri)

# shows updated image in image viewer
plt.imshow(img_brightened)
```

# **Catatan**

Penyesuaian Kontras (Contrast): Parameter new_con ditetapkan pada angka 4. Pengaturan ini melonjakkan perbedaan intensitas warna, sehingga garis batas antara area terang dan gelap pada hamparan pohon cabe terlihat sangat kontras dan jelas.

```python
curr_con = ImageEnhance.Contrast(image)
new_con = 4

# Contrast enhanced by a factor of 0.3
img_contrasted = curr_con.enhance(new_con)

# shows updated image in image viewer
plt.imshow(img_contrasted)
```

# **Catatan**

Penyesuaian Ketajaman (Sharpness): Nilai parameter new_sharp disetel sebesar 3. Hasilnya, detail visual semakin tajam, di mana garis tepi dan tekstur pada pohon cabe tampak lebih kontras, tegas, serta terdefinisi dengan jelas.

```python
# Enhance Sharpness
curr_sharp = ImageEnhance.Sharpness(image)
new_sharp = 3

# Sharpness enhanced by a factor of 8.3
img_sharped = curr_sharp.enhance(new_sharp)

# shows updated image in image viewer
plt.imshow(img_sharped)
```

# **Catatan**

Penyesuaian Warna (Color): Parameter new_col ditetapkan pada angka 5. Pengaturan ini melonjakkan tingkat saturasi secara signifikan, sehingga warna hijau pada pohon cabe tampak jauh lebih pekat dan mencolok.

```python
curr_col = ImageEnhance.Color(image)
new_col = 5

# Color level enhanced by a factor of 2.5
img_colored = curr_col.enhance(new_col)

# shows updated image in image viewer
plt.imshow(img_colored )
```
